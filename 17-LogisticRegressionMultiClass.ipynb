```python
# src_packet_rate -> Source-side packet transmission rate
# dst_packet_rate -> Destination-side packet reception rate
# avg_payload_size -> Average size of payload in packets
# connection_duration -> Duration of the connection (in seconds)
# tcp_flag_count -> Number of TCP flag occurrences
# avg_interarrival_time -> Time between packet arrivals
# failed_login_attempts -> Number of failed login attempts
# unusual_port_activity_score -> Score representing unusual port usage
# session_entropy -> Entropy of session behavior (for anomaly detection)
# avg_response_delay -> Average delay in server response (in ms)
# attack_type -> 0 = Normal, 1 = DDoS, 2 = Port Scan
#------------------------------------------------------------------------
# src_packet_rate -> Kaynak taraf paket iletim hızı
# dst_packet_rate -> Hedef taraf paket alım hızı
# avg_payload_size -> Paketlerdeki ortalama yük boyutu
# connection_duration -> Bağlantı süresi (saniye cinsinden)
# tcp_flag_count -> TCP bayrağı oluşum sayısı
# avg_interarrival_time -> Paket gelişleri arasındaki süre
# failed_login_attempts -> Başarısız giriş deneme sayısı
# unusual_port_activity_score -> Olağandışı port kullanımını temsil eden puan
# session_entropy -> Oturum davranışının entropisi (anomali tespiti için)
# avg_response_delay -> Sunucu yanıtındaki ortalama gecikme (ms cinsinden)
# attack_type -> 0 = Normal, 1 = DDoS, 2 = Port Tarama
```

```python
X = df.drop("attack_type", axis=1) # attack_type haricini al X e at
y = df["attack_type"] # attack_type i y ye atar
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.3, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
score = accuracy_score(y_pred, y_test)
print("score: ", score)
print("clasiification_report: \n", classification_report(y_pred,y_test))
print("confusion matrix: \n", confusion_matrix(y_pred,y_test))
```

```python
from sklearn.multiclass import OneVsOneClassifier, OneVsRestClassifier
```

```python
onevsonemodel = OneVsOneClassifier(LogisticRegression())
```

```python
onevsrestmodel = OneVsRestClassifier(LogisticRegression())
```

```python
onevsonemodel.fit(X_train, y_train)
y_pred = onevsonemodel.predict(X_test)
score = accuracy_score(y_pred, y_test)
print("score: ", score)
print("clasiification_report: \n", classification_report(y_pred,y_test))
print("confusion matrix: \n", confusion_matrix(y_pred,y_test))
```

```python
onevsrestmodel.fit(X_train, y_train)
y_pred = onevsrestmodel.predict(X_test)
score = accuracy_score(y_pred, y_test)
print("score: ", score)
print("clasiification_report: \n", classification_report(y_pred,y_test))
print("confusion matrix: \n", confusion_matrix(y_pred,y_test))
```
