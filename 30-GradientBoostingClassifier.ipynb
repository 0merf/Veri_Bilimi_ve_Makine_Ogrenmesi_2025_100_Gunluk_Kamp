```python
X = df.drop('target', axis = 1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 15)
```

```python
def correlation_for_dropping(df, threshold):
    columns_to_drop = set() # set tanimliyoruz, bunu icinde olacak silinecekler
    corr = df.corr()
    for i in range(len(corr.columns)): # bir kolon ile diger tum kolonlari tek tek karsilastiriyor
        for j in range(i):
            if abs(corr.iloc[i, j]) > threshold:
                colname = corr.columns[i]
                columns_to_drop.add(colname)
    return columns_to_drop
```

```python
correlation_for_dropping(X_train, 0.85)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.model_selection import GridSearchCV
```

```python
gb = GradientBoostingClassifier()
gb.fit(X_train, y_train)
```

```python
y_pred = gb.predict(X_test)
print(accuracy_score(y_pred, y_test))
print(classification_report(y_pred, y_test))
print(confusion_matrix(y_pred, y_test))
```

```python
parameters = {
        "loss" : ['log_loss', 'exponential'],
        "learning_rate" : [0.01, 0.05, 0.1],
        "n_estimators" : [100, 150, 180, 200],
        "max_depth" : [3,4,5],
        "subsample" : [0.8, 1.0]
}
```

```python
grid_search = GridSearchCV(estimator=GradientBoostingClassifier(), param_grid = parameters, cv = 5, n_jobs = -1, verbose=1)
```

```python
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
y_pred = grid_search.predict(X_test)
print(accuracy_score(y_pred, y_test))
print(classification_report(y_pred, y_test))
print(confusion_matrix(y_pred, y_test))
```
